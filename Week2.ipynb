```python
img = io.imread('Mobil.jpg')

img2 = cv.cvtColor(img, cv.COLOR_BGR2RGB)

R, G, B = cv.split(img2)

hitam = np.zeros_like(R)

img_RB = cv.merge([R, hitam, B])
img_GB = cv.merge([hitam, G, B])

plt.figure(figsize=(10, 5))
plt.suptitle("Kombinasi Kanal Warna", fontsize=16)

plt.subplot(1, 2, 1)
plt.imshow(img_RB)
plt.title("Kombinasi Merah - Biru")

plt.subplot(1, 2, 2)
plt.imshow(img_GB)
plt.title("Kombinasi Hijau - Biru")

plt.show()
```

```python
img_vertikal = cv.flip(img2, 0)
img_horizontal = cv.flip(img2, 1)
img_dua = cv.flip(img2, -1)

plt.figure(figsize=(10, 5))
plt.suptitle("Pencerminan Citra", fontsize=16)

plt.subplot(1, 3, 1)
plt.imshow(img_vertikal)
plt.title("Pencerminan Vertikal")

plt.subplot(1, 3, 2)
plt.imshow(img_horizontal)
plt.title("Pencerminan Horizontal")

plt.subplot(1, 3, 3)
plt.imshow(img_dua)
plt.title("Pencerminan Vertikal dan Horizontal")

plt.show()
```

```python
def label_adaptif(citra, teks="Dokumentasi PCVK - 2026"):

    hasil = citra.copy()
    h, w = hasil.shape[:2]

    bar_h = int(h / 12)

    hasil[h - bar_h:h, :] = 0

    font = cv.FONT_HERSHEY_SIMPLEX
    font_scale = bar_h / 45.0
    thickness = max(1, int(bar_h / 25))


    pos_x = int(w * 0.02)
    pos_y = h - int(bar_h * 0.35)


    cv.putText(hasil, teks, (pos_x, pos_y), font, font_scale, (255, 255, 255), thickness, cv.LINE_AA)

    return hasil

img = cv.imread('WhatsApp Image 2026-09-08 at 13.23.38.jpeg')
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)
hasil1 = (label_adaptif(img))

img2_resized = cv.resize(img_rgb, (480, 760))
hasil2 = label_adaptif(img2_resized)

# Menampilkan hasil
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title(f"Ukuran: {hasil1.shape[1]}x{hasil1.shape[0]}")
plt.imshow(hasil1)
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title(f"Ukuran: {hasil2.shape[1]}x{hasil2.shape[0]}")
plt.imshow(hasil2)
plt.axis('off')

plt.show()
```

```python
def buat_thumbnail(citra, nama_toko="TOKO RESMI"):
    h, w = citra.shape[:2]


    sisi = max(h, w)


    kanvas = np.full((sisi, sisi, 3), 255, dtype=np.uint8)


    start_y = (sisi - h) // 2
    start_x = (sisi - w) // 2


    kanvas[start_y:start_y + h, start_x:start_x + w] = citra


    font = cv.FONT_HERSHEY_SIMPLEX
    font_scale = sisi / 600.0
    thickness = max(1, int(sisi / 300))

    pos_x = int(sisi * 0.05)
    pos_y = int(sisi * 0.08)

    cv.putText(kanvas, nama_toko, (pos_x, pos_y), font, font_scale, (50, 50, 50), thickness, cv.LINE_AA)

    return kanvas


img = cv.imread('WhatsApp Image 2026-09-08 at 13.23.38.jpeg')
img_rgb = cv.cvtColor(img, cv.COLOR_BGR2RGB)


hasil_thumbnail = buat_thumbnail(img_rgb)


plt.figure(figsize=(6, 6))
plt.imshow(hasil_thumbnail)
plt.title(f"Thumbnail Persegi: {hasil_thumbnail.shape[1]}x{hasil_thumbnail.shape[0]}")
plt.axis('off')
plt.show()
```
