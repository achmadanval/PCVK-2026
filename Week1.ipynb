```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

### Langakh 2

Baca imgae dari URLs - Read image From URls

Pada langkah ini kita akan emmbaca image dari URLs, menampilkanya menggunakan OpenCV, Perhatikan ada perbedaan ketika membuka imgae dalam RGB dan BGR, Channel warna default untuk OpenCV dalah BGR.

```python
# Membuat List untuk menyimpan URL dari beberapa imgae
imageurls =[
    "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/fruits.jpg",
    "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/butterfly.jpg",
    "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/starry_night.jpg",
]# Baca dan tampilkan image
# loop pada tiap url image, bebrapa image dapat disimpan pada list

for url in imageurls:
  image = io.imread(url)
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5)
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
  final_frame = np.concatenate((image, image_2))
  cv2_imshow(final_frame)
  print('\n')

```

### Langkah 3

```python
tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("Resolusi image: Tinggi x lebar = ",tinggi," x ",lebar)
cv2_imshow(image_2)
```

```python
image_3 = image.copy()

row = int(image_3.shape[0] / 2)

#garis horizontal di tengah image tanpa loop
#image_3[int(image.shape[0]/2),:]=[0,255,0]

#membuat garis horizontal ditengah image dengan loop

for x in range(image_3.shape[1]):
  image_3[row, x] = [0, 255, 0]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```
